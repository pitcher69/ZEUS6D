```python
def make_gif_from_folder(image_folder, output_gif_path):
    """
    Create an animated GIF from a folder of images.
    
    Args:
        image_folder (str): Path to directory with images.
        output_gif_path (str): Path to save output GIF.
    """
    import glob
    from PIL import Image

    image_files = sorted(glob.glob(f"{image_folder}/*.png"))
    frames = []

    for img_path in image_files:
        img = Image.open(img_path).convert("RGBA")
        bg = Image.new("RGB", img.size, (0, 0, 0))
        bg.paste(img, mask=img.split()[3])
        frames.append(bg)

    if frames:
        frames[0].save(
            output_gif_path,
            format="GIF",
            save_all=True,
            append_images=frames[1:],
            duration=150,
            loop=0
        )
        print(f"✅ Saved GIF to: {output_gif_path}")
    else:
        print(f"⚠️ No images found in {image_folder}, GIF not created.")
```

```python
import os
import cv2
import torch
import numpy as np
from PIL import Image
from tqdm import tqdm
import requests
from segment_anything import sam_model_registry, SamAutomaticMaskGenerator
from transformers import AutoImageProcessor, AutoModelForDepthEstimation

device = "cuda" if torch.cuda.is_available() else "cpu"
video_id = "video50"
frames_dir = video_id
rgb_frames = sorted([f for f in os.listdir(frames_dir) if f.endswith(".png")])

# === Output folders ===
masks_root = f"{video_id}_masks"
depth_root = f"{video_id}_depth"
os.makedirs(masks_root, exist_ok=True)
os.makedirs(depth_root, exist_ok=True)

# === Load SAM ViT-H ===
SAM_CHECKPOINT = "sam_vit_h_4b8939.pth"
if not os.path.exists(SAM_CHECKPOINT):
    print("📥 Downloading SAM checkpoint...")
    url = "https://dl.fbaipublicfiles.com/segment_anything/sam_vit_h_4b8939.pth"
    r = requests.get(url, stream=True)
    with open(SAM_CHECKPOINT, "wb") as f:
        for chunk in r.iter_content(8192):
            f.write(chunk)

print("🎯 Loading SAM model...")
sam = sam_model_registry["vit_h"](checkpoint=SAM_CHECKPOINT).to(device)
mask_gen = SamAutomaticMaskGenerator(sam)

# === Load Depth Anything with use_fast=True ===
print("🌊 Loading Depth Anything V2 Large...")
depth_processor = AutoImageProcessor.from_pretrained(
    "depth-anything/Depth-Anything-V2-Large-hf",
    use_fast=True
)
depth_model = AutoModelForDepthEstimation.from_pretrained("depth-anything/Depth-Anything-V2-Large-hf")
depth_model.to(device).eval()

# === Process all frames ===
print(f"📸 Processing frames in {video_id}...")
for frame_name in tqdm(rgb_frames, desc="🧠 Segment + Depth"):
    frame_id = os.path.splitext(frame_name)[0]
    frame_path = os.path.join(frames_dir, frame_name)

    # Load image
    scene_bgr = cv2.imread(frame_path)
    scene_rgb = cv2.cvtColor(scene_bgr, cv2.COLOR_BGR2RGB)
    scene_pil = Image.fromarray(scene_rgb)

    # --- Generate masks ---
    masks = mask_gen.generate(scene_rgb)
    masks = sorted(masks, key=lambda m: m["area"], reverse=True)

    mask_out_dir = os.path.join(masks_root, f"{frame_id}_masks")
    os.makedirs(mask_out_dir, exist_ok=True)

    for i, mask in enumerate(masks):
        binary_mask = np.zeros(scene_rgb.shape[:2], dtype=np.uint8)
        binary_mask[mask["segmentation"]] = 255
        mask_path = os.path.join(mask_out_dir, f"mask_{i+1:02d}.png")
        cv2.imwrite(mask_path, binary_mask)

    # --- Generate depth ---
    inputs = depth_processor(images=scene_pil, return_tensors="pt").to(device)
    with torch.no_grad():
        outputs = depth_model(**inputs)

    post_processed = depth_processor.post_process_depth_estimation(
        outputs,
        target_sizes=[(scene_pil.height, scene_pil.width)],
    )
    depth_tensor = post_processed[0]["predicted_depth"]
    depth_np = depth_tensor.detach().cpu().numpy()

    # Normalize depth to 0–255 for uint8 image
    depth_vis = (depth_np - depth_np.min()) / (depth_np.max() - depth_np.min() + 1e-8)
    depth_vis_uint8 = (depth_vis * 255).astype(np.uint8)

    # Save normalized depth image
    depth_out_path = os.path.join(depth_root, f"{frame_id}_depth.png")
    cv2.imwrite(depth_out_path, depth_vis_uint8)

print("✅ All masks and normalized depth maps saved.")
```

```python
import os
import torch
import numpy as np
from PIL import Image
from tqdm import tqdm
from sklearn.metrics.pairwise import cosine_similarity
from transformers import AutoImageProcessor, AutoModel

# === Setup ===
device = "mps" if torch.mps.is_available() else "cpu"
video_id = "video50"
model_dir = "models"
renders_dir = "renders"
frames_dir = video_id
all_masks_root = f"{video_id}_masks"
output_root = f"{video_id}_mask"
os.makedirs(output_root, exist_ok=True)

# === Load model ===
print("🔄 Loading DINOv2-Base model...")
processor = AutoImageProcessor.from_pretrained("facebook/dinov2-base")
model = AutoModel.from_pretrained("facebook/dinov2-base").to(device)
model.eval()

# === Feature Extraction ===
def extract_dino_feature(image_or_path, cls_token=True):
    if isinstance(image_or_path, str):
        image = Image.open(image_or_path).convert("RGB")
    else:
        image = image_or_path.convert("RGB")
    inputs = processor(images=image, return_tensors="pt").to(device)
    with torch.no_grad():
        outputs = model(**inputs)
    feat = outputs.last_hidden_state[:, 0, :] if cls_token else outputs.last_hidden_state.mean(dim=1)
    return feat.squeeze().cpu().numpy()

# === Load all object render features ===
object_render_features = {}
object_ids = [os.path.splitext(f)[0] for f in os.listdir(model_dir) if f.endswith(".ply")]

for obj_id in tqdm(object_ids, desc="🧠 Extracting Object Render Features"):
    render_path = os.path.join(renders_dir, f"{obj_id}_render")
    render_imgs = sorted([f for f in os.listdir(render_path) if f.endswith(".png")])
    feats = []
    for img_file in render_imgs:
        img_path = os.path.join(render_path, img_file)
        feats.append(extract_dino_feature(img_path))
    object_render_features[obj_id] = feats

# === Match each object in each frame ===
frame_ids = sorted([f[:-4] for f in os.listdir(frames_dir) if f.endswith(".png")])

for frame_id in tqdm(frame_ids, desc="🖼️ Processing Frames"):
    frame_path = os.path.join(frames_dir, f"{frame_id}.png")
    scene_rgb = Image.open(frame_path).convert("RGB")
    scene_np = np.array(scene_rgb)
    masks_dir = os.path.join(all_masks_root, f"{frame_id}_masks")
    mask_files = sorted([f for f in os.listdir(masks_dir) if f.endswith(".png")])

    # Precompute all mask features
    mask_features = []
    for mask_file in mask_files:
        mask_path = os.path.join(masks_dir, mask_file)
        mask = Image.open(mask_path).convert("L")
        mask_np = np.array(mask) > 128
        if not np.any(mask_np):
            continue
        mean_rgb = scene_np[mask_np].mean(axis=0)
        crop = scene_np.copy()
        crop[~mask_np] = mean_rgb.astype(np.uint8)
        crop_img = Image.fromarray(crop)
        feat = extract_dino_feature(crop_img)
        mask_features.append((mask_file, feat))

    # For each object, find best matching mask
    for obj_id, render_feats in object_render_features.items():
        best_match = (None, -1)
        for mask_file, feat in mask_features:
            best_sim = max(
                cosine_similarity(view_feat.reshape(1, -1), feat.reshape(1, -1))[0][0]
                for view_feat in render_feats
            )
            if best_sim > best_match[1]:
                best_match = (mask_file, best_sim)

        best_mask_file, best_score = best_match
        if best_score < 0.45:
            print(f"⚠️ Frame {frame_id}, Object {obj_id}: low similarity ({best_score:.3f})")
            continue

        # Create subdirectory for object
        obj_out_dir = os.path.join(output_root, obj_id)
        os.makedirs(obj_out_dir, exist_ok=True)

        mask_in = os.path.join(masks_dir, best_mask_file)
        mask_out = os.path.join(obj_out_dir, f"{frame_id}_{obj_id}.png")
        Image.open(mask_in).save(mask_out)
        print(f"✅ Frame {frame_id}, Object {obj_id}: saved {mask_out} (sim={best_score:.3f})")
```

```python
import os
import glob
from PIL import Image
import numpy as np

video_id = "video50"
mask_root = f"{video_id}_mask"  # e.g., video50_mask/000000/
rgb_root = "video50"  # e.g., video50/rgb/
output_gif_root = "."  # Current directory

object_ids = [d for d in os.listdir(mask_root) if os.path.isdir(os.path.join(mask_root, d))]

def make_overlay_gif(rgb_dir, mask_dir, output_gif_path):
    mask_files = sorted(glob.glob(os.path.join(mask_dir, "*.png")))
    frames = []

    for mask_path in mask_files:
        # Extract frame number from filename
        filename = os.path.basename(mask_path)
        frame_id = filename.split("_")[0]  # e.g., 000620
        rgb_path = os.path.join(rgb_dir, f"{frame_id}.png")

        if not os.path.exists(rgb_path):
            print(f"⚠️ Missing RGB frame: {rgb_path}")
            continue

        # Open and prepare images
        rgb_img = Image.open(rgb_path).convert("RGBA")
        mask_img = Image.open(mask_path).convert("L")

        # Convert mask to red overlay with transparency
        red_mask = Image.fromarray(np.zeros((mask_img.height, mask_img.width, 4), dtype=np.uint8))
        red_mask_np = np.array(red_mask)
        mask_np = np.array(mask_img)
        red_mask_np[mask_np > 0] = [255, 0, 0, 100]  # Semi-transparent red
        red_mask = Image.fromarray(red_mask_np, mode='RGBA')

        # Composite the overlay
        composite = Image.alpha_composite(rgb_img, red_mask)
        frames.append(composite.convert("RGB"))

    if frames:
        frames[0].save(
            output_gif_path,
            format="GIF",
            save_all=True,
            append_images=frames[1:],
            duration=150,
            loop=0
        )
        print(f"✅ Saved overlay GIF: {output_gif_path}")
    else:
        print(f"⚠️ No valid frames in {mask_dir}, skipping GIF.")

# Run for each object
for obj_id in object_ids:
    mask_dir = os.path.join(mask_root, obj_id)
    output_gif = os.path.join(output_gif_root, f"{video_id}_{obj_id}_overlay.gif")
    make_overlay_gif(rgb_root, mask_dir, output_gif)
```
